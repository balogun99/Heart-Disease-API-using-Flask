# Import Libraries

```python
# import libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import pickle
```

# Load the dataset

```python
# load the dataset
data = pd.read_csv('/Users/macbookpro/Desktop/Heart Disease API Using Flask/heart_disease_data.csv')
```

```python
# print first 5 rows of the dataset
data.head()
```

```python
# drop the target column
X = data.drop(columns=['target'])
y = data['target']
```

```python
# independent 
X
```

```python
# dependent
y
```

# Data Splitting

```python
# split the dataset into training & testing dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
# know the shape of the spliited data
print(y.shape, y_train.shape, y_test.shape)
```

#### 303 => Data Points
#### 242 => Training Data
#### 61 => Testing Data

```python
# scale the features using Standard Scaler
scaler = StandardScaler()
```

```python
# scale the train data
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Model Training

```python
# train the model
model = LogisticRegression()
```

```python
# fit the model based on scaled data
model.fit(X_train_scaled, y_train) # y => Target
```

# Model Prediction

```python
# prediction on training data
y_train_pred = model.predict(X_train_scaled)
accuracy = accuracy_score(y_train, y_train_pred) # y_test => True Labels ; y_pred => predicted labels
print("Training accuracy: ", round(accuracy))
```

```python
# prediction on test data
y_pred = model.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred) # y_test => True Labels ; y_pred => predicted labels
print("Testing accuracy: ", round(accuracy))
```

# Save the Models

#### Scaler & Pickle File

```python
# model file
with open('heart_disease_model.pkl', 'wb') as model_file:
    pickle.dump(model, model_file)

# scaler file
with open('heart_disease_scaler.pkl', 'wb') as scaler_file:
    pickle.dump(scaler, scaler_file)
```
